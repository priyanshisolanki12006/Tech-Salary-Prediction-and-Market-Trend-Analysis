## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# 1. Create copy of dataset
df1 = df.copy()

# 2. Check shape before cleaning
print("Shape Before Cleaning:", df1.shape)

# 3. Replace hidden missing values
df1.replace(['-1',
             'Unknown',
             'Unknown / Non-Applicable'],
             np.nan,
             inplace=True)

# 4. Check missing values
print("\nMissing Values Before Handling:")
print(df1.isnull().sum())

# 5. Remove duplicate values
df1.drop_duplicates(inplace=True)

# 6. Handle missing values in categorical columns
categorical_cols = df1.select_dtypes(include='object').columns

for col in categorical_cols:
    df1[col].fillna('Not Available', inplace=True)

# 7. Handle missing values in numerical columns
numerical_cols = df1.select_dtypes(include=np.number).columns

for col in numerical_cols:
    df1[col].fillna(df1[col].median(), inplace=True)

# 8. Clean column names
df1.columns = df1.columns.str.strip()
df1.columns = df1.columns.str.lower()
df1.columns = df1.columns.str.replace(" ", "_")

# 9. Check shape after cleaning
print("\nShape After Cleaning:", df1.shape)

# 10. Final missing values check
print("\nTotal Missing Values After Cleaning:",
      df1.isnull().sum().sum())

# Final dataset preview
df1.head()
````

### What all manipulations have you done and insights you found?

Data Wrangling Insights:

1. A copy of the original dataset was created to avoid modifications to the raw data.
2. Hidden missing values such as '-1', 'Unknown', and 'Unknown / Non-Applicable' were replaced with null values for accurate preprocessing.
3. Duplicate rows were identified and removed to improve data quality.
4. Missing values in categorical columns were handled by replacing them with 'Not Available'.
5. Missing values in numerical columns were filled using median values to minimize the effect of outliers.
6. Column names were standardized by converting them to lowercase and replacing spaces with underscores for consistency.
7. After preprocessing, the dataset became clean, structured, and ready for Exploratory Data Analysis (EDA) and Machine Learning modeling.
8. The salary-related features, job title, location, company size, and industry appear to be important variables for prediction and trend analysis.

**Salary Feature Engineering**

````python
# ==============================
# Salary Feature Engineering (Fixed Final)
# ==============================

# Remove missing values
df1 = df1[df1['salary_estimate'] != 'Not Available']

# Extract salary ranges
salary = df1['salary_estimate'].str.extract(r'(\d+)\D*-\D*(\d+)')

# Remove rows where salary extraction failed
salary = salary.dropna()

# Keep same rows in dataframe
df1 = df1.loc[salary.index]

# Convert to integer
df1['min_salary'] = salary[0].astype(int)
df1['max_salary'] = salary[1].astype(int)

# Create average salary
df1['avg_salary'] = (
    df1['min_salary'] +
    df1['max_salary']
) / 2

# Check output
df1[['salary_estimate',
     'min_salary',
     'max_salary',
     'avg_salary']].head()
````

## ***6. Feature Engineering & Data Pre-processing***

### 1. Handling Missing Values

````python
# Check Missing Values
df1.isnull().sum()

# Handling categorical missing values

categorical_cols = df1.select_dtypes(include='object').columns

for col in categorical_cols:
    df1[col].fillna('Not Available', inplace=True)

# Handling numerical missing values

numerical_cols = df1.select_dtypes(include=['int64', 'float64']).columns

for col in numerical_cols:
    df1[col].fillna(df1[col].median(), inplace=True)

# Verify missing values
df1.isnull().sum().sum()
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

Techniques Used:
Categorical Data → Mode/Custom Category (Not Available)
Numerical Data → Median Imputation

Why?
Median is robust to outliers.
"Not Available" preserves categorical information.

### 2. Handling Outliers

Before Outlier Treatment

````python
plt.figure(figsize=(10,5))

sns.boxplot(df1['avg_salary'])

plt.title('Outliers in Average Salary')
plt.show()
````

Outlier Treatment Using IQR Method

````python
# IQR Method

Q1 = df1['avg_salary'].quantile(0.25)
Q3 = df1['avg_salary'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

df1 = df1[
    (df1['avg_salary'] >= lower_limit) &
    (df1['avg_salary'] <= upper_limit)
]
````

After Outlier Treatment

````python
plt.figure(figsize=(10,5))

sns.boxplot(df1['avg_salary'])

plt.title('After Outlier Treatment')

plt.show()
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

Technique Used: IQR (Interquartile Range)

Why?
Because salary data is often skewed and IQR works effectively without assuming normal distribution.

### ML Model - 1 (Linear Regression)

Importing Required Libraries

````python
# Machine Learning Libraries
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.model_selection import (
    cross_val_score,
    GridSearchCV
)

import numpy as np
````

Fit the Algorithm

````python
# Linear Regression Model

lr_model = LinearRegression()

# Train Model
lr_model.fit(X_train, y_train)
````

Predict on the Model

Evaluation Metrics

Evaluation Metric Score Chart

````python
# Score Chart

scores = {
    'MAE': mae_lr,
    'RMSE': rmse_lr,
    'R2 Score': r2_lr
}

plt.figure(figsize=(8,5))

plt.bar(scores.keys(),
        scores.values())

plt.title('Linear Regression Performance')

plt.show()
````

Predict After Optimization
